```python
import pandas as pd

df = pd.read_csv("car_risk.csv")

df.head()
```

```python
print(df.isnull().sum())
```

```python
median_value = df["experience"].median()
df["experience"] = df["experience"].fillna(median_value)
print("Median:", median_value)

display(df)
```

```python

from sklearn.linear_model import LinearRegression

# Independent Variables (Features)
X = df[["speed", "car_age", "experience"]]

# Dependent Variable (Target)
y = df["risk"]

# Create Model
model = LinearRegression()

# Train Model Relationship
model.fit(X, y)

print("Model Training Completed!")
```

```python
print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
# Get coefficients
print("Coefficients:", model.coef_)

# Get intercept
print("Intercept:", model.intercept_)
```

```python
print("Speed Coefficient:", model.coef_[0])
print("Car Age Coefficient:", model.coef_[1])
print("Experience Coefficient:", model.coef_[2])
```

```python
# New Input Data
new_data = pd.DataFrame({
    "speed": [160],
    "car_age": [10],
    "experience": [5]
})

# Predict Risk
prediction = model.predict(new_data)

print("Predicted Risk:", prediction[0])
```

```python
new_data2 = pd.DataFrame({
    "speed": [120],
    "car_age": [8],
    "experience": [10]
})

prediction2 = model.predict(new_data2)

print("Predicted Risk:", prediction2[0])
```

```python
# Predict training dataset
y_pred = model.predict(X)

# Compare actual and predicted values
comparison = pd.DataFrame({
    "Actual Risk": y,
    "Predicted Risk": y_pred
})

print(comparison.round(2))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("R2 Score:", round(r2, 4))
```
